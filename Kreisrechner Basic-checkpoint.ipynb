```python
import math

def kreisflaeche(radius):
    return math.pi * radius ** 2

try:
    radius = float(input("Bitte gib den Radius ein: ").replace(",", "."))
    if radius < 0:
        print("Der Radius darf nicht negativ sein.")
    else:
        flaeche = kreisflaeche(radius)
        print(f"Die Fläche des Kreises beträgt: {flaeche:.2f}")
except ValueError:
    print("Ungültige Eingabe. Bitte gib eine Zahl ein.")
```
