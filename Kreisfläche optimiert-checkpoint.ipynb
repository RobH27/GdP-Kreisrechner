```python
"""
Kreisflächen-Rechner mit Einheiten

Berechnet die Fläche eines Kreises anhand des eingegebenen Radius
und gibt das Ergebnis in der passenden Quadrat-Einheit aus.
Beispiel-Eingabe: 5 cm
"""

import math


def kreisflaeche(radius):
    """Berechnet die Fläche eines Kreises (π · r²)."""
    return math.pi * radius ** 2


try:
    eingabe = input("Bitte gib den Radius mit Einheit ein (z. B. 5 cm): ")
    teile = eingabe.replace(",", ".").split()

    zahl = float(teile[0])
    einheit = teile[1] if len(teile) > 1 else ""

    if zahl < 0:
        print("Der Radius darf nicht negativ sein.")
    else:
        flaeche = kreisflaeche(zahl)
        print(f"Die Fläche des Kreises beträgt: {flaeche:.2f} {einheit}²")
except (ValueError, IndexError):
    print("Ungültige Eingabe. Bitte gib z. B. '5 cm' ein.")
```
