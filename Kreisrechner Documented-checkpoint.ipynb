```python
"""
Kreisflächen-Rechner

Berechnet die Fläche eines Kreises anhand des vom Benutzer
eingegebenen Radius (Formel: A = π · r²).
"""

import math


def kreisflaeche(radius):
    """
    Berechnet die Fläche eines Kreises.

    Parameter:
        radius (float): Der Radius des Kreises. Muss >= 0 sein.

    Rückgabe:
        float: Die Fläche des Kreises (π · radius²).

    Beispiel:
        >>> round(kreisflaeche(5), 2)
        78.54
    """
    return math.pi * radius ** 2


try:
    radius = float(input("Bitte gib den Radius ein: ").replace(",", "."))
    if radius < 0:
        print("Der Radius darf nicht negativ sein.")
    else:
        flaeche = kreisflaeche(radius)
        print(f"Die Fläche des Kreises beträgt: {flaeche:.3f}")
except ValueError:
    print("Ungültige Eingabe. Bitte gib eine Zahl ein.")
```
