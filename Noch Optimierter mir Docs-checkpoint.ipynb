```python
"""
Kreisflächen-Rechner mit Einheiten und Umrechnung

Erlaubte Eingaben: "5", "5 cm", "5cm", "2,5 m"
Unterstützte Einheiten: mm, cm, dm, m, km
"""

import math
import re

# Wie viele Meter ist 1 der jeweiligen Einheit?
FAKTOREN = {"mm": 0.001, "cm": 0.01, "dm": 0.1, "m": 1, "km": 1000}


def kreisflaeche(radius):
    """Berechnet die Fläche eines Kreises (π · r²)."""
    return math.pi * radius ** 2


def eingabe_zerlegen(text):
    """
    Zerlegt einen Text wie '5cm' in Zahl und Einheit.

    Rückgabe:
        tuple: (zahl, einheit). Fehlt etwas, steht dort None.
    Raises:
        ValueError: Wenn der Text nicht passt.
    """
    text = text.replace(",", ".").strip()
    treffer = re.fullmatch(r"(-?\d+(?:\.\d+)?)?\s*([a-zA-Z]+)?", text)
    if not treffer or text == "":
        raise ValueError("Ungültige Eingabe.")
    zahl_text, einheit = treffer.groups()
    zahl = float(zahl_text) if zahl_text else None
    return zahl, einheit.lower() if einheit else None


def main():
    while True:
        try:
            zahl, einheit = eingabe_zerlegen(
                input("Radius eingeben (z. B. 5, 5 cm oder 5cm): ")
            )
        except ValueError:
            print("Ungültige Eingabe. Beispiele: 5 | 5 cm | 5cm | 2,5 m")
            continue

        if zahl is None:
            print(f"Du hast nur die Einheit '{einheit}' eingegeben. Es fehlt die Zahl.")
            continue
        if zahl < 0:
            print("Der Radius darf nicht negativ sein.")
            continue
        if einheit is not None and einheit not in FAKTOREN:
            print(f"Unbekannte Einheit '{einheit}'. Erlaubt: {', '.join(FAKTOREN)}")
            continue
        break

    flaeche = kreisflaeche(zahl)

    # Ohne Einheit: nur die Zahl ausgeben
    if einheit is None:
        print(f"Die Fläche des Kreises beträgt: {flaeche:.2f}")
        return

    # Mit Einheit: optional in andere Einheit umrechnen
    while True:
        ziel = input(f"Ergebnis in welcher Einheit? (Enter = {einheit}²): ").strip().lower()
        if ziel == "":
            ziel = einheit
        if ziel in FAKTOREN:
            break
        print(f"Unbekannte Einheit. Erlaubt: {', '.join(FAKTOREN)}")

    # Umrechnung: erst in m², dann in die Zieleinheit
    flaeche_m2 = flaeche * FAKTOREN[einheit] ** 2
    ergebnis = flaeche_m2 / FAKTOREN[ziel] ** 2
    print(f"Die Fläche des Kreises beträgt: {ergebnis:.2f} {ziel}²")

main()
```
